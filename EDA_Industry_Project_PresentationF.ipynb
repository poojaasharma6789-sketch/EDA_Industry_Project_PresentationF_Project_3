```python
# ==========================================
# 📊 FINAL MASTER PROJECT: E
# (Fixed and Optimized Version)
# ==========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew, kurtosis  # Statistical analysis ke liye

# Plot style setup
sns.set_theme(style="whitegrid")
plt.rcParams.update({
    'font.size': 11,
    'axes.labelsize': 12,
    'axes.titlesize': 14,
    'figure.titlesize': 16
})

# 1. Dataset Load karna (With Proper Error Handling)
try:
    df = pd.read_csv("INDUSTRY.csv")
    print("✅ Dataset successfully load!")
    print(f"Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}\n")
    
    # --- DATA CLEANING & QUALITY CHECK (this is in right place) ---
    print("📌 Missing Values Check:")
    print(df.isnull().sum())
    
    print("\n📌 Duplicate Rows Count:", df.duplicated().sum())
    if df.duplicated().sum() > 0:
        print("⚠️ Warning: Dataset me duplicate records hain.")
    else:
        print("✅ No duplicate records found.")
    print("\n" + "="*50 + "\n")

    # 2. HIGH-LEVEL STATISTICAL REPORT
    print("--- 📑 NUMERICAL FEATURES STATISTICAL REPORT ---")
    num_cols = ['employee_count', 'annual_revenue_million', 'profit_margin_percent', 'founded_year', 'customer_count', 'market_rating']

    stats_df = df[num_cols].describe().T
    stats_df['skewness'] = df[num_cols].skew()
    stats_df['kurtosis'] = df[num_cols].kurt()
    print(stats_df.round(2).to_string())
    print("\n" + "="*50 + "\n")

    # 3. VISUALIZATION GENERATION + AUTO SAVE

    # --- CHART 1: Revenue Distribution ---
    print("📊 Generating Chart 1: Revenue Distribution...")
    plt.figure(figsize=(10, 5))
    sns.histplot(df['annual_revenue_million'], kde=True, color='#3498db', bins=40)
    plt.title('Univariate Analysis: Distribution of Annual Revenue ($ Millions)', pad=15)
    plt.xlabel('Annual Revenue (Millions)')
    plt.ylabel('Frequency')
    plt.tight_layout()
    plt.savefig('1_univariate_revenue.png', dpi=300)
    plt.show()

    # --- CHART 2: Correlation Matrix Heatmap ---
    print("📊 Generating Chart 2: Correlation Matrix Heatmap...")
    plt.figure(figsize=(8, 6))
    corr = df[num_cols].corr()
    sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
    plt.title('Correlation Matrix: Checking Feature Relationships', pad=15)
    plt.tight_layout()
    plt.savefig('2_correlation_matrix.png', dpi=300)
    plt.show()

    # --- CHART 3: Employee Count vs Annual Revenue ---
    print("📊 Generating Chart 3: Employee Count vs Annual Revenue...")
    plt.figure(figsize=(10, 6))
    sns.scatterplot(data=df, x='employee_count', y='annual_revenue_million', alpha=0.6, hue='industry', palette='Set2')
    plt.title('Bivariate Analysis: Employee Count vs Annual Revenue by Sector', pad=15)
    plt.xlabel('Total Employee Count')
    plt.ylabel('Annual Revenue (Millions)')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title='Industry')
    plt.tight_layout()
    plt.savefig('3_bivariate_analysis.png', dpi=300)
    plt.show()

    # --- CHART 4: Sector-wise Performance ---
    print("📊 Generating Chart 4: Sector-wise Performance...")
    plt.figure(figsize=(11, 5))
    sns.barplot(data=df, x='industry', y='annual_revenue_million', estimator=np.mean, errorbar=None, palette='muted')
    plt.title('Multivariate Analysis: Average Annual Revenue across Industries', pad=15)
    plt.xlabel('Industry Sector')
    plt.ylabel('Average Revenue ($ Millions)')
    plt.xticks(rotation=15)
    plt.tight_layout()
    plt.savefig('4_multivariate_industry_performance.png', dpi=300)
    plt.show()

    # --- CHART 5: Outlier Detection ---
    print("📊 Generating Chart 5: Outlier Detection (Boxplot)...")
    plt.figure(figsize=(10, 5))
    sns.boxplot(x=df['annual_revenue_million'], color='#e74c3c')
    plt.title('Outlier Detection - Annual Revenue', pad=15)
    plt.xlabel('Annual Revenue (Millions)')
    plt.tight_layout()
    plt.savefig('5_outlier_detection.png', dpi=300)
    plt.show()

    print("\n🏁 PERFECT! All the report data has been printed successfully, and all five charts have been saved in your folder!")

except FileNotFoundError:
    print("❌ Error: 'INDUSTRY.csv' The file could not be found. Please make sure it is located in the same folder as your Python script or Jupyter Notebook.")
except Exception as e:
    print(f"❌ An unknown error occurred: {e}")
```
